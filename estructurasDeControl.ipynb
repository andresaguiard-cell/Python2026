### 1  Estructuras de control

**1.1 Selección (sentencias condicionales)**

* Selección de una de varias alternativas en base a alguna condición.
* Indentación para estructurar el código
* Importante el símbolo ‘:’


```python
# # Asigno el bvalo uno a la variable x
x = int(input("Ingrese el valor de x"))
#Implemento una condicional para saber si x es mayor a 0
if x > 0:
   print('Valor positivo')
else:
   print('Valor no positivo')
```

* Las condiciones son expresiones que se evalúan a True o False.
* Operadores de comparación, lógicos, de identidad y de pertenencia vistos anteriormente.

```python
x = 3
y = [1, 2, 3]

print(x > 0)
print(x > 0 and x < 10)
print(x is not y)
print(x in y)
```

* Se pueden introducir más ‘ramas’ en sentencias condicionales a través de la palabra clave *elif*.

```python
x = -3

if x > 0:
   print('Valor positivo')
elif x == 0:
   print('Valor nulo')
else:
   print('Valor negativo')
```

* Anidamiento.

```python
val = 120
if val > 0:
    if val < 100:
        print('Valor positivo')
    else:
        print('Valor muy positivo')
else:
    print('Valor negativo')
```

2. Escribir un programa en python que pida al usuario un  numero entero por teclado y muestre en pantalla si es par o impar.

```python
n = int(input("Ingrese un número entero: "))

if n % 2 == 0:
    print("El número es par.")
else:
    print("El número es impar.")
```

3. Escribir un programa en python para una empresa que tiene salas de juego para todas las edades y quiere calcular de forma automática el precio que debe cobrar asus clientes por entrada.El programa debe preguntar al usuario: el nombre y la edad del cliente y mostrar el precio de entreda si el cliennte es menor de 4 años puede ingresar gratis, si tiene entre 4 y 18 años debe pagar 5$ y si es mayor de 18 años 10$

```python
nombre = input("nombre del cliente: ")
edad = int(input("edad del cliente: "))

if edad < 4:
    print(f"El cliente {nombre} puede entrar GRATIS.")
elif 4 <= edad <= 18:
    print(f"El cliente {nombre} debe pagar $5.")
else:
    print(f"El cliente {nombre} debe pagar $10.")
```
